```python
# Task 1: While Loop & Control Statements -Number Guessing Game
# Task 1: While Loop & Control Statements - Number Guessing Game

## 🟢 Random Number
Generates a random number between 1 and 10 using `random.randint()`.

## 🟢 While Loop
The `while` loop allows the user to make a maximum of 3 attempts.

## 🟢 Check the Guess
`if`, `elif`, and `else` check whether the guess is out of range, correct, too high, or too low.

## 🟢 Break
`break` stops the loop when the correct number is guessed.

## 🟢 Attempt Counter
`i += 1` increases the attempt count after each guess.

## 🟢 While-Else
The `else` block displays **"Better luck next time"** if all 3 attempts are unsuccessful.
```

```python
import random

secret_number = random.randint(1, 10)
attempt = 3
i = 1

while i <= 3:
    guess = int(input("Enter the guess number between 1 to 10: "))

    if guess < 1 or guess > 10:
        print("Your guess is out of range")

    elif guess == secret_number:
        print("Congrats! You guessed the correct number")
        break

    elif guess > secret_number:
        print("Too high, try again")

    else:
        print("Too low, try again")

    i += 1

else:
    print("Better luck next time")
```

```python
#Task 2:- For Loop - Multiplication Table Generator
## 🟢 Input
The user enters a number to generate its multiplication table.

## 🟢 Range
`range(1, 11)` generates numbers from 1 to 10.

## 🟢 Calculation
Each number is multiplied by `i` to get the multiplication result.

## 🟢 Output
The table is printed using an f-string in the format `number * i = result`.
```

```python
number = int(input("Enter the number for which you want the multiplication table: "))
print(f"multiplication table for {number}")
for i in range(1, 11):
    result = number * i
    print(f"{number} * {i} = {result}")
```

```python
# Task 3: Function - BMI Calculator

## 🟢 Define Function
The `calculate_bmi()` function takes `weight` and `height` as inputs.

## 🟢 Calculate BMI
The BMI is calculated using the formula `weight / (height ** 2)`.

## 🟢 User Input
The user enters weight in kilograms and height in meters.

## 🟢 Function Call
The function is called with the given weight and height to calculate the BMI.

## 🟢 Display Result
The calculated BMI is displayed using an f-string.
```

```python
def calculate_bmi(weight,height):
   return weight/(height**2)

weight = int(input("enter the weight in kg:"))
height = float(input("enter the height in metre:"))
bmi = calculate_bmi(weight,height)
print(f"your bmi is:{bmi}")
```
